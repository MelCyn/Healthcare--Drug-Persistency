## **Outlier Detection**
Moving on to **detecting outliers** in the data which may cause the data to be skewed. In this case, we'll check for outliers in numerical columns using the **IQR** (Inter-Quartile Range) method and decide on how to handle them depending is being observed.

Considering numeric columns that aren't binary (i.e not just 0/1) based on the data summary.

```python
# Helper function to detect outliers using IQR
def detect_outliers_iqr(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"{col} ➤ Outliers detected: {outliers.shape[0]}")
    print(f"  ▪ Range: {df[col].min()} to {df[col].max()}")
    print(f"  ▪ IQR Boundaries: {lower_bound:.2f} to {upper_bound:.2f}")
    print("-" * 50)
    return outliers

# Check outliers in a few columns
for col in ['Dexa_Freq_During_Rx', 'Gluco_Record_Prior_Ntm', 'Gluco_Record_During_Rx']:
    detect_outliers_iqr(df_cleaned, col)
```

Below gives the **Outlier Summary**



| Column                   | Outliers | Min → Max | IQR Boundaries  | Notes                                 |
| ------------------------ | -------- | --------- | --------------- | ------------------------------------- |
| `Dexa_Freq_During_Rx`    | **460**  | 0 → 146   | **-4.5 to 7.5** | Clearly has high outliers (e.g., 146) |
| `Gluco_Record_Prior_Ntm` | **805**  | 0 → 1     | **0 to 0**      | Binary (0/1); IQR not meaningful      |
| `Gluco_Record_During_Rx` | **0**    | 0 → 1     | -1.5 to 2.5     | No outliers                           |

**Interpretation**

*   Dexa_Freq_During_Rx: Frequency > 7.5 is considered an outlier — this makes sense for a frequency column.
*   Gluco_Record_Prior_Ntm and Gluco_Record_During_Rx: These are binary indicators (0/1), so IQR isn't appropriate. The 805 flagged are just all values equal to 1 — not true outliers.

To handle this, lets consider capping the outliers by setting them to the maximum acceptable limit (ie at the upper IQR bound [7.5]).

```python
# Cap values in Dexa_Freq_During_Rx to 7.5 (IQR upper bound)
upper_cap = df_cleaned['Dexa_Freq_During_Rx'].quantile(0.75) + 1.5 * (df_cleaned['Dexa_Freq_During_Rx'].quantile(0.75) - df_cleaned['Dexa_Freq_During_Rx'].quantile(0.25))
df_cleaned['Dexa_Freq_During_Rx_Capped'] = df_cleaned['Dexa_Freq_During_Rx'].clip(upper=upper_cap)
```

```python
# Compare min, max, mean, etc.
print("Original Dexa_Freq_During_Rx:")
print(df['Dexa_Freq_During_Rx'].describe())

print("\nCapped Dexa_Freq_During_Rx_Capped:")
print(df_cleaned['Dexa_Freq_During_Rx_Capped'].describe())
```

**Summary of Results**

*   The mean and standard deviation dropped significantly — this is expected, as extreme values (like 146) can skew the mean and inflate the spread.
*    Maximum reduced from 146 → 7.5 — this confirms the cap was applied correctly.
*   No unintended loss of data — the count remains 3424, so no rows were dropped.

```python
# Set the upper limit used during capping
upper_bound = 7.5  # from previous IQR analysis

# Count how many were above that before capping
num_capped = (df['Dexa_Freq_During_Rx'] > upper_bound).sum()
print(f"Number of values capped in Dexa_Freq_During_Rx: {num_capped}")
```
